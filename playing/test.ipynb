```python
import random

population_size = 10
chromosome_length = 5
mutation_rate = 0.1
generation = 1

def fitness_function(x):
    return (-x**2)/10 + 3*x

def generate_chromosome():
    return [random.randint(0,1) for i in range(chromosome_length)]
    
def decode_chromosome(chromosome):

    binary_string = ""
    for bit in chromosome:
        binary_string += str(bit)

    return int(binary_string, 2)

def crossover(parent1, parent2):
    point = random.randint(1, chromosome_length-1)

    child1 = parent1[:point] + parent2[point:]
    child2 = parent2[:point] + parent1[point:]

    return child1, child2

def mutate(chromosome):
    for i in range(chromosome_length):
        if random.random() < mutation_rate:

            chromosome[i] = 1 - chromosome[i]
    return chromosome

population = [generate_chromosome() for i in range(population_size)]

while True:
    print("\nGeneration: ",generation)

    fitness_values = []
    for chromosome in population:
        x = decode_chromosome(chromosome)
        fitness = fitness_function(x)
        fitness_values.append(fitness)
        print("Chromosome: ", chromosome, "x =", x, "fitness =", round(fitness,2))

    best_fitness = max(fitness_values)
    best_index = fitness_values.index(best_fitness)
    best_chromosome = population[best_index]
    best_x = decode_chromosome(best_chromosome)

    if best_fitness >= 22:
        print("\nBest Solution Found!")
        print("Best Chromosome: ", best_chromosome)
        print("Best x:", best_x)
        print("Best Fitness:", round(best_fitness,2))

        break

    selected = random.choices(population, weights = fitness_values, k = population_size)

    new_population = []

    for i in range(0, population_size, 2):

        parent1 = selected[i]
        parent2 = selected[i+1]

        child1, child2 = crossover(parent1, parent2)

        if generation % 3 == 0:
            child1 = mutate(child1)
            child2 = mutate(child2)

        new_population.extend([child1, child2])

    population = new_population
    generation += 1
    
```

```python
import random

cities = ['A', 'B', 'C', 'D', 'E']

distances = {
    ('A','B'):10, ('A','C'):15, ('A','D'):20, ('A','E'):25,
    ('B','A'):10, ('B','C'):35, ('B','D'):25, ('B','E'):17,
    ('C','A'):15, ('C','B'):35, ('C','D'):30, ('C','E'):28,
    ('D','A'):20, ('D','B'):25, ('D','C'):30, ('D','E'):12,
    ('E','A'):25, ('E','B'):17, ('E','C'):28, ('E','D'):12
}

population_size = 6
generations = 20
mutation_rate = 0.1

def generate_route():
    route = cities[1:]

    random.shuffle(route)
    return route

def total_distance(route):
    path = ['A'] + route + ['A']
    distance = 0
    for i in range(len(path)-1):
        distance += distances[(path[i], path[i+1])]

    return distance

def fitness(route):
    return 1/total_distance(route)

def crossover(parent1, parent2):

    point = random.randint(1, len(parent1)-1)
    child = parent1[:point]

    for city in parent2:
        if city not in child:
            child.append(city)

    return child

def mutate(route):
    if random.random() < mutation_rate:
        i = random.randint(0,len(route)-1)
        j = random.randint(0,len(route)-1)

        route[i], route[j] = route[j], route[i]



population = []

for i in range(population_size):
    population.append(generate_route())

## MAIN LOOP
for generation in range(generations):
    print("\nGeneration", generation+1)
    fitness_values = []
    for route in population:
        fit = fitness(route)
        fitness_values.append(fit)
        print(route, "Distance = ", total_distance(route))

    new_population = []
    for i in range(population_size):
        parent1=random.choices(population, weights = fitness_values)[0]
        parent2=random.choices(population, weights = fitness_values)[0]

        child = crossover(parent1, parent2)
        mutate(child)
        new_population.append(child)
    population = new_population

best_route = min(population, key = total_distance)
print("\nBest Route Found: ")
print("A -> "+ " -> ".join(best_route) + " -> A")
print("total distance =", total_distance(best_route))



```
