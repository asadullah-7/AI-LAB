# DLS

```python
def DLS(graph, start, goal, depth, path, cost):
    print(f"Current node {start} , depth is {depth} cost is  {cost}")

    if start == goal:
        print(f"Goal found! path {path} , cost is {cost}")
        return True

    if depth == 0:
        return False
    
    for i,w in graph[start].items():
        found = DLS(graph,i,goal,depth-1,path+[i],cost+w)
        if found :
            return True
    return False
    
graph = {
    'A': {'B': 2, 'C': 3},
    'B': {'D': 4, 'E': 1},
    'C': {'F': 5},
    'D': {},
    'E': {},
    'F': {}
}

DLS(graph,'A','E',3,['A'],0)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.naive_bayes import GaussianNB

file = "diabetes.csv"


data = pd.read_csv(file)

X = data.iloc[:,:-1]
Y = data.iloc[:,-1]

X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.33, random_state = 42)

model = GaussianNB()
model.fit(X_train,Y_train)

prediction = model.predict(X_test)

accuracy = accuracy_score(Y_test, prediction)*100
print (accuracy)


```

# MiniMax

```python
import math
import random

board = [' ' for i in range(9)]

def print_board():
    print()
    for i in range(3):
        print(board[i*3] + " | " + board[i*3+1] + " | " + board[i*3+2])
        if i < 2:
            print("--+---+--")
    print()


def check_winner(b,player):
    win_states = [
        [0,1,2],[3,4,5],[6,7,8],
        [0,3,6],[1,4,7],[2,5,8],
        [0,4,8],[2,4,6]
    ]

    for state in win_states:
        if all(b[i] == player for i in state):
            return True
    return False

def is_draw(b):
    return ' ' not in b

def get_moves(b):
    return [i for i in range(9) if b[i] == ' ']

def heuristic(b):
    score = 0
    lines = [
        [0,1,2],[3,4,5],[6,7,8],
        [0,3,6],[1,4,7],[2,5,8],
        [0,4,8],[2,4,6]
    ]

    for line in lines:
        x_count = sum(1 for i in line if b[i] == 'X')
        o_count = sum(1 for i in line if b[i] == 'O')


        if x_count > 0 and o_count == 0:
            score += x_count
        elif o_count > 0 and x_count == 0:
            score -= o_count
    
    return score

def minimax(b, depth, is_max, max_depth = 3):

    if check_winner(b, 'X'):
        return 10-depth
    if check_winner(b, 'O'):
        return depth-10
    if is_draw(b):
        return 0
    
    if depth == max_depth:
        return heuristic(b)

    moves = get_moves(b)

    if is_max:
        best = -math.inf
        for move in moves:
            b[move] = 'X' 
            val = minimax(b, depth+1, False, max_depth)
            b[move] = ' '
            best = max(best, val)

        return best
    
    else:
        best = math.inf
        for move in moves:
            b[move] = 'O'
            val = minimax(b, depth+1, True, max_depth)
            b[move] = ' '
            best = min(best, val)

        return best
    
def best_move():
    
    moves = get_moves(board)
    best_score = -math.inf
    best_moves = []

    for move in moves:
        board[move] = 'X'
        score = minimax(board, 0, False)
        board[move] = ' '

        if score > best_score:
            best_score = score
            best_moves = [move]
        elif score == best_score:
            best_moves.append(move)

    return random.choice(best_moves)

def human_move():
    moves = get_moves(board)
    move = random.choice(moves)  # Random (as required)
    board[move] = 'O'
    print(f"Human chose position {move}")

# Game loop
def play_game():
    print("Tic Tac Toe (Computer = X, Human = O)")
    print_board()
    
    while True:
        # Computer move
        move = best_move()
        board[move] = 'X'
        print(f"Computer chose position {move}")
        print_board()
        
        if check_winner(board, 'X'):
            print("Computer wins!")
            break
        if is_draw(board):
            print("Draw!")
            break
        
        # Human move
        human_move()
        print_board()
        
        if check_winner(board, 'O'):
            print("Human wins!")
            break
        if is_draw(board):
            print("Draw!")
            break

play_game()
```

# Generic Algorithm

```python

import random

# -----------------------------
# PARAMETERS
# -----------------------------
population_size = 10
chromosome_length = 5
mutation_rate = 0.1
generation = 1

# -----------------------------
# FITNESS FUNCTION
# -----------------------------
def fitness_function(x):
    return (-x**2)/10 + 3*x

# -----------------------------
# GENERATE RANDOM CHROMOSOME
# -----------------------------
def generate_chromosome():
    return [random.randint(0,1) for _ in range(chromosome_length)]

# -----------------------------
# DECODE BINARY TO DECIMAL
# -----------------------------
def decode_chromosome(chromosome):

    binary_string = ""

    for bit in chromosome:
        binary_string += str(bit)

    return int(binary_string, 2)

# -----------------------------
# CROSSOVER
# -----------------------------
def crossover(parent1, parent2):

    point = random.randint(1, chromosome_length-1)

    child1 = parent1[:point] + parent2[point:]
    child2 = parent2[:point] + parent1[point:]

    return child1, child2

# -----------------------------
# MUTATION
# -----------------------------
def mutate(chromosome):
 
    for i in range(chromosome_length):

        if random.random() < mutation_rate:

            chromosome[i] = 1 - chromosome[i]

    return chromosome

# -----------------------------
# INITIAL POPULATION
# -----------------------------
population = [generate_chromosome() for _ in range(population_size)]

# -----------------------------
# MAIN LOOP
# -----------------------------
while True:

    print("\nGeneration:", generation)

    # Calculate fitness
    fitness_values = []

    for chromosome in population:

        x = decode_chromosome(chromosome)

        fitness = fitness_function(x)

        fitness_values.append(fitness)

        print("Chromosome:", chromosome, "x =", x,"fitness =", round(fitness,2))

    # Best chromosome
    best_fitness = max(fitness_values)

    best_index = fitness_values.index(best_fitness)

    best_chromosome = population[best_index]

    best_x = decode_chromosome(best_chromosome)

    # TERMINATION CONDITION
    if best_fitness >= 22:

        print("\nBest Solution Found!")
        print("Best Chromosome:", best_chromosome)
        print("Best x:", best_x)
        print("Best Fitness:", round(best_fitness,2))

        break

    # -----------------------------
    # SELECTION (Roulette Wheel)
    # -----------------------------
    selected = random.choices(population,weights=fitness_values,k=population_size)

    # -----------------------------
    # CREATE NEW POPULATION
    # -----------------------------
    new_population = []

    for i in range(0, population_size, 2):

        parent1 = selected[i]
        parent2 = selected[i+1]

        child1, child2 = crossover(parent1, parent2)

        # Mutation every 3 generations
        if generation % 3 == 0:

            child1 = mutate(child1)
            child2 = mutate(child2)

        new_population.extend([child1, child2])

    population = new_population

    generation += 1
```

# TSP

```python
import random

# --------------------------------
# CITIES
# --------------------------------
cities = ['A', 'B', 'C', 'D', 'E']

# --------------------------------
# DISTANCE MATRIX
# --------------------------------
distances = {
    ('A','B'):10, ('A','C'):15, ('A','D'):20, ('A','E'):25,
    ('B','A'):10, ('B','C'):35, ('B','D'):25, ('B','E'):17,
    ('C','A'):15, ('C','B'):35, ('C','D'):30, ('C','E'):28,
    ('D','A'):20, ('D','B'):25, ('D','C'):30, ('D','E'):12,
    ('E','A'):25, ('E','B'):17, ('E','C'):28, ('E','D'):12
}

# --------------------------------
# PARAMETERS
# --------------------------------
population_size = 6
generations = 20
mutation_rate = 0.1

# --------------------------------
# GENERATE RANDOM ROUTE
# --------------------------------
def generate_route():

    route = cities[1:]   # exclude A

    random.shuffle(route)

    return route

# --------------------------------
# TOTAL DISTANCE
# --------------------------------
def total_distance(route):

    path = ['A'] + route + ['A']

    distance = 0

    for i in range(len(path)-1):

        distance += distances[(path[i], path[i+1])]

    return distance

# --------------------------------
# FITNESS FUNCTION
# --------------------------------
def fitness(route):

    return 1 / total_distance(route)

# --------------------------------
# CROSSOVER
# --------------------------------
def crossover(parent1, parent2):

    point = random.randint(1, len(parent1)-1)

    child = parent1[:point]

    for city in parent2:

        if city not in child: # Duplication avoid kry ga....

            child.append(city)

    return child

# --------------------------------
# MUTATION
# --------------------------------
def mutate(route):

    if random.random() < mutation_rate:

        i = random.randint(0, len(route)-1)
        j = random.randint(0, len(route)-1)

        route[i], route[j] = route[j], route[i]

# --------------------------------
# INITIAL POPULATION
# --------------------------------
population = []

for _ in range(population_size):

    population.append(generate_route())

# --------------------------------
# MAIN LOOP
# --------------------------------
for generation in range(generations):

    print("\nGeneration", generation+1)

    fitness_values = []

    for route in population:

        fit = fitness(route)

        fitness_values.append(fit)

        print(route, "Distance =", total_distance(route))

    # NEW POPULATION
    new_population = []

    for _ in range(population_size):

        parent1 = random.choices(population,weights=fitness_values)[0]

        parent2 = random.choices(population,weights=fitness_values)[0]

        child = crossover(parent1, parent2)

        mutate(child)

        new_population.append(child)

    population = new_population

# --------------------------------
# BEST ROUTE
# --------------------------------
best_route = min(population, key=total_distance)

print("\nBest Route Found:")

print("A -> " + " -> ".join(best_route) + " -> A")

print("Total Distance =", total_distance(best_route))
```
