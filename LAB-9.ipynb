```python

import random
import math

# Board setup
board = [' ' for _ in range(9)]

# Print board
def print_board():
    print()
    for i in range(3):
        print(board[i*3] + " | " + board[i*3+1] + " | " + board[i*3+2])
        if i < 2:
            print("--+---+--")
    print()

# Check winner
def check_winner(b, player):
    win_states = [
        [0,1,2], [3,4,5], [6,7,8],  # rows
        [0,3,6], [1,4,7], [2,5,8],  # cols
        [0,4,8], [2,4,6]            # diagonals
    ]
    for state in win_states:
        if all(b[i] == player for i in state):
            return True
    return False

# Check draw
def is_draw(b):
    return ' ' not in b

# Get available moves
def get_moves(b):
    return [i for i in range(9) if b[i] == ' ']

# Heuristic function (Exercise 8.4)
def heuristic(b):
    score = 0
    lines = [
        [0,1,2],[3,4,5],[6,7,8],
        [0,3,6],[1,4,7],[2,5,8],
        [0,4,8],[2,4,6]
    ]
    
    for line in lines:
        x_count = sum(1 for i in line if b[i] == 'X')
        o_count = sum(1 for i in line if b[i] == 'O')
        
        if x_count > 0 and o_count == 0:
            score += x_count
        elif o_count > 0 and x_count == 0:
            score -= o_count
    
    return score

def minimax(b, depth, is_max, alpha, beta, max_depth=3):
    
    if check_winner(b, 'X'):
        return 10 - depth
    if check_winner(b, 'O'):
        return depth - 10
    if is_draw(b):
        return 0

    if depth == max_depth:
        return heuristic(b)

    moves = get_moves(b)

    if is_max:
        best = -math.inf

        for move in moves:
            b[move] = 'X'
            val = minimax(b, depth + 1, False, alpha, beta, max_depth)
            b[move] = ' '
            
            best = max(best, val)
            alpha = max(alpha, best)

            #PRUNING
            if beta <= alpha:
                break

        return best

    else:
        best = math.inf

        for move in moves:
            b[move] = 'O'
            val = minimax(b, depth + 1, True, alpha, beta, max_depth)
            b[move] = ' '
            
            best = min(best, val)
            beta = min(beta, best)

            # PRUNING
            if beta <= alpha:
                break

        return best

# Get best move with random tie-break
def best_move():
    moves = get_moves(board)
    best_score = -math.inf
    best_moves = []
    
    for move in moves:
        board[move] = 'X'
        score = minimax(board, 0, False, -math.inf, math.inf)
        board[move] = ' '
        
        if score > best_score:
            best_score = score
            best_moves = [move]
        elif score == best_score:
            best_moves.append(move)
    
    return random.choice(best_moves)

# Human move
def human_move():
    moves = get_moves(board)
    move = random.choice(moves)  # Random (as required)
    board[move] = 'O'
    print(f"Human chose position {move}")

# Game loop
def play_game():
    print("Tic Tac Toe (Computer = X, Human = O)")
    print_board()
    
    while True:
        # Computer move
        move = best_move()
        board[move] = 'X'
        print(f"Computer chose position {move}")
        print_board()
        
        if check_winner(board, 'X'):
            print("Computer wins!")
            break
        if is_draw(board):
            print("Draw!")
            break
        
        # Human move
        human_move()
        print_board()
        
        if check_winner(board, 'O'):
            print("Human wins!")
            break
        if is_draw(board):
            print("Draw!")
            break

# Run game
play_game()

 
```

```python

import random

# -----------------------------
# PARAMETERS
# -----------------------------
population_size = 10
chromosome_length = 5
mutation_rate = 0.1
generation = 1

# -----------------------------
# FITNESS FUNCTION
# -----------------------------
def fitness_function(x):
    return (-x**2)/10 + 3*x

# -----------------------------
# GENERATE RANDOM CHROMOSOME
# -----------------------------
def generate_chromosome():
    return [random.randint(0,1) for _ in range(chromosome_length)]

# -----------------------------
# DECODE BINARY TO DECIMAL
# -----------------------------
def decode_chromosome(chromosome):

    binary_string = ""

    for bit in chromosome:
        binary_string += str(bit)

    return int(binary_string, 2)

# -----------------------------
# CROSSOVER
# -----------------------------
def crossover(parent1, parent2):

    point = random.randint(1, chromosome_length-1)

    child1 = parent1[:point] + parent2[point:]
    child2 = parent2[:point] + parent1[point:]

    return child1, child2

# -----------------------------
# MUTATION
# -----------------------------
def mutate(chromosome):
 
    for i in range(chromosome_length):

        if random.random() < mutation_rate:

            chromosome[i] = 1 - chromosome[i]

    return chromosome

# -----------------------------
# INITIAL POPULATION
# -----------------------------
population = [generate_chromosome() for _ in range(population_size)]

# -----------------------------
# MAIN LOOP
# -----------------------------
while True:

    print("\nGeneration:", generation)

    # Calculate fitness
    fitness_values = []

    for chromosome in population:

        x = decode_chromosome(chromosome)

        fitness = fitness_function(x)

        fitness_values.append(fitness)

        print("Chromosome:", chromosome, "x =", x,"fitness =", round(fitness,2))

    # Best chromosome
    best_fitness = max(fitness_values)

    best_index = fitness_values.index(best_fitness)

    best_chromosome = population[best_index]

    best_x = decode_chromosome(best_chromosome)

    # TERMINATION CONDITION
    if best_fitness >= 22:

        print("\nBest Solution Found!")
        print("Best Chromosome:", best_chromosome)
        print("Best x:", best_x)
        print("Best Fitness:", round(best_fitness,2))

        break

    # -----------------------------
    # SELECTION (Roulette Wheel)
    # -----------------------------
    selected = random.choices(population,weights=fitness_values,k=population_size)

    # -----------------------------
    # CREATE NEW POPULATION
    # -----------------------------
    new_population = []

    for i in range(0, population_size, 2):

        parent1 = selected[i]
        parent2 = selected[i+1]

        child1, child2 = crossover(parent1, parent2)

        # Mutation every 3 generations
        if generation % 3 == 0:

            child1 = mutate(child1)
            child2 = mutate(child2)

        new_population.extend([child1, child2])

    population = new_population

    generation += 1
```
