```python

import random

# -----------------------------
# PARAMETERS
# -----------------------------
population_size = 10
chromosome_length = 5
mutation_rate = 0.1
generation = 1

# -----------------------------
# FITNESS FUNCTION
# -----------------------------
def fitness_function(x):
    return (-x**2)/10 + 3*x

# -----------------------------
# GENERATE RANDOM CHROMOSOME
# -----------------------------
def generate_chromosome():
    return [random.randint(0,1) for _ in range(chromosome_length)]

# -----------------------------
# DECODE BINARY TO DECIMAL
# -----------------------------
def decode_chromosome(chromosome):

    binary_string = ""

    for bit in chromosome:
        binary_string += str(bit)

    return int(binary_string, 2)

# -----------------------------
# CROSSOVER
# -----------------------------
def crossover(parent1, parent2):

    point = random.randint(1, chromosome_length-1)

    child1 = parent1[:point] + parent2[point:]
    child2 = parent2[:point] + parent1[point:]

    return child1, child2

# -----------------------------
# MUTATION
# -----------------------------
def mutate(chromosome):
 
    for i in range(chromosome_length):

        if random.random() < mutation_rate:

            chromosome[i] = 1 - chromosome[i]

    return chromosome

# -----------------------------
# INITIAL POPULATION
# -----------------------------
population = [generate_chromosome() for _ in range(population_size)]

# -----------------------------
# MAIN LOOP
# -----------------------------
while True:

    print("\nGeneration:", generation)

    # Calculate fitness
    fitness_values = []

    for chromosome in population:

        x = decode_chromosome(chromosome)

        fitness = fitness_function(x)

        fitness_values.append(fitness)

        print("Chromosome:", chromosome, "x =", x,"fitness =", round(fitness,2))

    # Best chromosome
    best_fitness = max(fitness_values)

    best_index = fitness_values.index(best_fitness)

    best_chromosome = population[best_index]

    best_x = decode_chromosome(best_chromosome)

    # TERMINATION CONDITION
    if best_fitness >= 22:

        print("\nBest Solution Found!")
        print("Best Chromosome:", best_chromosome)
        print("Best x:", best_x)
        print("Best Fitness:", round(best_fitness,2))

        break

    # -----------------------------
    # SELECTION (Roulette Wheel)
    # -----------------------------
    selected = random.choices(population,weights=fitness_values,k=population_size)

    # -----------------------------
    # CREATE NEW POPULATION
    # -----------------------------
    new_population = []

    for i in range(0, population_size, 2):

        parent1 = selected[i]
        parent2 = selected[i+1]

        child1, child2 = crossover(parent1, parent2)

        # Mutation every 3 generations
        if generation % 3 == 0:

            child1 = mutate(child1)
            child2 = mutate(child2)

        new_population.extend([child1, child2])

    population = new_population

    generation += 1
```

```python
import random

# --------------------------------
# CITIES
# --------------------------------
cities = ['A', 'B', 'C', 'D', 'E']

# --------------------------------
# DISTANCE MATRIX
# --------------------------------
distances = {
    ('A','B'):10, ('A','C'):15, ('A','D'):20, ('A','E'):25,
    ('B','A'):10, ('B','C'):35, ('B','D'):25, ('B','E'):17,
    ('C','A'):15, ('C','B'):35, ('C','D'):30, ('C','E'):28,
    ('D','A'):20, ('D','B'):25, ('D','C'):30, ('D','E'):12,
    ('E','A'):25, ('E','B'):17, ('E','C'):28, ('E','D'):12
}

# --------------------------------
# PARAMETERS
# --------------------------------
population_size = 6
generations = 20
mutation_rate = 0.1

# --------------------------------
# GENERATE RANDOM ROUTE
# --------------------------------
def generate_route():

    route = cities[1:]   # exclude A

    random.shuffle(route)

    return route

# --------------------------------
# TOTAL DISTANCE
# --------------------------------
def total_distance(route):

    path = ['A'] + route + ['A']

    distance = 0

    for i in range(len(path)-1):

        distance += distances[(path[i], path[i+1])]

    return distance

# --------------------------------
# FITNESS FUNCTION
# --------------------------------
def fitness(route):

    return 1 / total_distance(route)

# --------------------------------
# CROSSOVER
# --------------------------------
def crossover(parent1, parent2):

    point = random.randint(1, len(parent1)-1)

    child = parent1[:point]

    for city in parent2:

        if city not in child: # Duplication avoid kry ga....

            child.append(city)

    return child

# --------------------------------
# MUTATION
# --------------------------------
def mutate(route):

    if random.random() < mutation_rate:

        i = random.randint(0, len(route)-1)
        j = random.randint(0, len(route)-1)

        route[i], route[j] = route[j], route[i]

# --------------------------------
# INITIAL POPULATION
# --------------------------------
population = []

for _ in range(population_size):

    population.append(generate_route())

# --------------------------------
# MAIN LOOP
# --------------------------------
for generation in range(generations):

    print("\nGeneration", generation+1)

    fitness_values = []

    for route in population:

        fit = fitness(route)

        fitness_values.append(fit)

        print(route, "Distance =", total_distance(route))

    # NEW POPULATION
    new_population = []

    for _ in range(population_size):

        parent1 = random.choices(population,weights=fitness_values)[0]

        parent2 = random.choices(population,weights=fitness_values)[0]

        child = crossover(parent1, parent2)

        mutate(child)

        new_population.append(child)

    population = new_population

# --------------------------------
# BEST ROUTE
# --------------------------------
best_route = min(population, key=total_distance)

print("\nBest Route Found:")

print("A -> " + " -> ".join(best_route) + " -> A")

print("Total Distance =", total_distance(best_route))
```

```python
import random

# ---------------------------
# PARAMETERS
# ---------------------------
N = 8
population_size = 20
generations = 100
mutation_rate = 0.1

# ---------------------------
# CREATE CHROMOSOME
# ---------------------------
def generate_chromosome():
    return [random.randint(1, N) for _ in range(N)]

# ---------------------------
# FITNESS FUNCTION
# ---------------------------
def fitness(chromosome):

    non_attacking = 0

    max_pairs = (N * (N - 1)) // 2

    for i in range(N):
        for j in range(i+1, N):

            if chromosome[i] != chromosome[j] and \
               abs(chromosome[i] - chromosome[j]) != abs(i - j):

                non_attacking += 1

    return non_attacking

# ---------------------------
# CROSSOVER
# ---------------------------
def crossover(p1, p2):

    point = random.randint(1, N-2)

    return p1[:point] + p2[point:]

# ---------------------------
# MUTATION
# ---------------------------
def mutate(chromosome):

    if random.random() < mutation_rate:

        i = random.randint(0, N-1)

        chromosome[i] = random.randint(1, N)

# ---------------------------
# INITIAL POPULATION
# ---------------------------
population = [generate_chromosome() for _ in range(population_size)]

max_fitness = (N*(N-1))//2

# ---------------------------
# MAIN LOOP
# ---------------------------
for generation in range(generations):

    fitness_values = [fitness(c) for c in population]

    best = max(population, key=fitness)

    best_fit = fitness(best)

    print("Generation", generation+1, "Best Fitness:", best_fit)

    # STOP CONDITION
    if best_fit == max_fitness:
        break

    # -----------------------
    # SELECTION
    # -----------------------
    selected = random.choices(
        population,
        weights=fitness_values,
        k=population_size
    )

    # -----------------------
    # NEW POPULATION
    # -----------------------
    new_population = []

    for i in range(population_size):

        p1 = selected[i]
        p2 = selected[(i+1) % population_size]

        child = crossover(p1, p2)

        mutate(child)

        new_population.append(child)

    population = new_population

# ---------------------------
# RESULT
# ---------------------------
best = max(population, key=fitness)

print("\nBEST SOLUTION:")
print(best)
print("Fitness:", fitness(best))
```
